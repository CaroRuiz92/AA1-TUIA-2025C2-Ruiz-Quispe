Se crea una columna para conocer bien los **KMs** hechos por viaje.

```python
# Distancia Haversine
def haversine(lat1, lon1, lat2, lon2):
    R = 6371
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1)*np.cos(lat2)*np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    return R * c
df_ny['KMs'] = haversine(df_ny['pickup_latitude'], df_ny['pickup_longitude'],
                      df_ny['dropoff_latitude'], df_ny['dropoff_longitude'])
```

```python
# Predicciones
y_train_pred = model.predict(X_train_scaled)
y_test_pred  = model.predict(X_test_scaled)

# Métricas
def evaluar_modelo(y_true, y_pred, nombre="modelo"):
    mse  = mean_squared_error(y_true, y_pred)
    rmse = mse**0.5
    mae  = mean_absolute_error(y_true, y_pred)
    r2   = r2_score(y_true, y_pred)
    print(f"Resultados {nombre}:")
    print(f"  MSE : {mse:.4f}")
    print(f"  RMSE: {rmse:.4f}")
    print(f"  MAE : {mae:.4f}")
    print(f"  R²  : {r2:.4f}")
    print("-"*40)

evaluar_modelo(y_train, y_train_pred, "LinearRegression (train)")
evaluar_modelo(y_test, y_test_pred, "LinearRegression (test)")
```

**Prueba con métodos de gradiente descendiente**

```python
from sklearn.linear_model import SGDRegressor
```

```python
m, n = X_train_scaled.shape

# Agregar columna de 1s 
X_train_bias = np.c_[np.ones((m,1)), X_train_scaled] 
X_test_bias = np.c_[np.ones((len(X_test_scaled),1)), X_test_scaled]

# Asegurar que y sean np.array columnas 
y_train = np.asarray(y_train).reshape(-1,1) 
y_test = np.asarray(y_test).reshape(-1,1)

# Inicialización 
theta_batch = np.random.randn(n+1,1) 
theta_sgd = np.random.randn(n+1,1) 
theta_mbgd = np.random.randn(n+1,1)

n_epochs = 50
eta = 0.1 
batch_size = 64 

errors = {"batch": [], "sgd": [], "minibatch": []}

for epoch in range(n_epochs): 
    # --- Batch GD --- 
    grad_batch = 2/m * X_train_bias.T.dot(X_train_bias.dot(theta_batch) - y_train) 
    theta_batch -= eta * grad_batch

    # --- SGD --- 
    i = np.random.randint(m) 
    xi = X_train_bias[i:i+1] 
    yi = y_train[i:i+1] 
    grad_sgd = 2 * xi.T.dot(xi.dot(theta_sgd) - yi) 
    theta_sgd -= eta * grad_sgd 

    # --- Mini-Batch --- 
    indices = np.random.randint(m, size=batch_size) 
    xi = X_train_bias[indices] 
    yi = y_train[indices] 
    grad_mbgd = 2/batch_size * xi.T.dot(xi.dot(theta_mbgd) - yi) 
    theta_mbgd -= eta * grad_mbgd

    # Errores en test 
    y_pred_batch = X_test_bias.dot(theta_batch) 
    y_pred_sgd = X_test_bias.dot(theta_sgd) 
    y_pred_mbgd = X_test_bias.dot(theta_mbgd) 
    
    errors["batch"].append(mean_squared_error(y_test, y_pred_batch)) 
    errors["sgd"].append(mean_squared_error(y_test, y_pred_sgd)) 
    errors["minibatch"].append(mean_squared_error(y_test, y_pred_mbgd))

# Graficar 
plt.plot(range(1, n_epochs+1), errors["batch"], label="Batch GD") 
plt.plot(range(1, n_epochs+1), errors["sgd"], label="SGD") 
plt.plot(range(1, n_epochs+1), errors["minibatch"], label="Mini-Batch GD") 
plt.xlabel("Épocas") 
plt.ylabel("Error (MSE en test)") 
plt.title("Comparación: Batch, SGD y Mini-Batch") 
plt.legend() 
plt.show()
```

**Observaciones:**

**Batch GD**: la curva es la más suave, lo que indica una convergencia estable y precisa. Su error desciende de manera constante hasta un valor mínimo.

**Mini-Batch GD**: esta curva es casi idéntica a la anterior, demostrando una convergencia igualmente estable y un rendimiento similar.

**SGD**: muestra un comportamiento errático y su error es mucho más alto. Esto se debe a que la tasa de aprendizaje es demasiado alta para este algoritmo o que el decaimiento no es el adecuado.

**Se prueba con métodos de regularización (Lasso, Ridge, Elastic Net)**

```python
# S define un rango amplio de alphas para probar
alphas = [0.001, 0.01, 0.1, 1, 10, 100]
resultados = []

for alpha in alphas:
    # Inicializa y entrena el modelo Ridge
    ridge_model = Ridge(alpha=alpha)
    ridge_model.fit(X_train, y_train)

    # Predicciones y métricas
    for conj, X, y in [('train', X_train, y_train), ('test', X_test, y_test)]:
        y_pred = ridge_model.predict(X)
        resultados.append({
            'Modelo': 'Ridge',
            'Alpha': alpha,
            'Conjunto': conj,
            'R²': r2_score(y, y_pred),
            'MSE': mean_squared_error(y, y_pred)
        })

df_resultados_ridge = pd.DataFrame(resultados)
print(df_resultados_ridge)
```

**Observaciones**: 

El R² en el conjunto de prueba se mantiene casi idéntico en 0.795547 para todos los valores de alpha probados, desde 0.001 hasta 100. Esto demuestra que la regularización L2 no tuvo un impacto significativo en el poder predictivo de tu modelo.

MSE (Error Cuadrático Medio): De manera similar, el MSE de prueba se mantiene en 18.395 para todo el rango de alpha. Esto confirma que la regularización no redujo el error del modelo.

```python
alphas = [0.001, 0.01, 0.1, 1, 10, 100]
resultados = []

for alpha in alphas:
    # Inicializa y entrena el modelo Lasso
    lasso_model = Lasso(alpha=alpha)
    lasso_model.fit(X_train, y_train)

    # Predicciones y métricas
    for conj, X, y in [('train', X_train, y_train), ('test', X_test, y_test)]:
        y_pred = lasso_model.predict(X)
        resultados.append({
            'Modelo': 'Lasso',
            'Alpha': alpha,
            'Conjunto': conj,
            'R²': r2_score(y, y_pred),
            'MSE': mean_squared_error(y, y_pred)
        })

df_resultados_lasso = pd.DataFrame(resultados)
print(df_resultados_lasso)
```

**Observaciones:** 

Alpha bajo (0.001 - 0.100): para valores bajos de alpha, el R² y MSE de Lasso son casi iguales a los de la regresión lineal estándar. Esto demuestra que una penalización pequeña no tiene un impacto significativo, ya que el modelo base ya es robusto.

Alpha medio (1.000): el R² de prueba cae a 0.7828, mientras que el MSE aumenta a 19.540. Esto indica que la penalización comienza a afectar el rendimiento del modelo, aunque todavía es aceptable.

Alpha alto (10.000 - 100.000): con valores altos de alpha, el R² baja a 0 o incluso a un valor negativo, y el MSE se dispara a 89.974. Esto significa que la penalización es tan fuerte que el modelo pierde toda su capacidad predictiva.

```python
alphas = [0.001, 0.01, 0.1, 1, 10, 100]
l1_ratios = [0.1, 0.5, 0.9] # Probar diferentes combinaciones de regularización
resultados = []

for alpha in alphas:
    for l1_ratio in l1_ratios:
        # Inicializa y entrena el modelo Elastic Net
        elastic_net_model = ElasticNet(alpha=alpha, l1_ratio=l1_ratio)
        elastic_net_model.fit(X_train, y_train)

        # Predicciones y métricas
        for conj, X, y in [('train', X_train, y_train), ('test', X_test, y_test)]:
            y_pred = elastic_net_model.predict(X)
            resultados.append({
                'Modelo': 'ElasticNet',
                'Alpha': alpha,
                'l1_ratio': l1_ratio,
                'Conjunto': conj,
                'R²': r2_score(y, y_pred),
                'MSE': mean_squared_error(y, y_pred)
            })

df_resultados_elastic_net = pd.DataFrame(resultados)
print(df_resultados_elastic_net)
```

**Observaciones**: 

Para valores de alpha de 0.001 y 0.01, el rendimiento de Elastic Net es casi idéntico al del modelo de regresión lineal estándar y al de Ridge. El R² de prueba se mantiene en 0.795, y el MSE en 18.395. Esto demuestra que, con una penalización mínima, el modelo no tiene un impacto significativo, ya que el modelo base es robusto.

Cuando alpha aumenta a 10 y 100, el R² se desploma a 0 o incluso a un valor negativo, y el MSE se dispara a 89.974. Esto significa que la penalización es tan fuerte que el modelo pierde toda su capacidad predictiva.

Influencia de l1_ratio: al variar l1_ratio, se puede ver cómo la combinación de regularizaciones afecta el rendimiento. A medida que alpha aumenta, el R² disminuye, sin importar la proporción entre L1 y L2. Esto confirma que una penalización excesiva es perjudicial para el modelo.

#### Optimización de hiperparámetros

**De gradiente descendiente**

```python
m, n = X_train_scaled.shape

# Agregar columna de 1s
X_train_bias = np.c_[np.ones((m, 1)), X_train_scaled]
X_test_bias = np.c_[np.ones((len(X_test_scaled), 1)), X_test_scaled]

# Asegurar que y sean np.array columnas
y_train = np.asarray(y_train).reshape(-1, 1)
y_test = np.asarray(y_test).reshape(-1, 1)

# Hiperparámetros
n_epochs = 50
batch_size = 64
etas = [0.1, 0.01, 0.001, 0.0001, 0.00001] # Diferentes learning rates
decay = 0.001 # para SGD con decaimiento

# Bucle para probar cada tasa de aprendizaje
for eta in etas:
    # Inicialización de parámetros
    theta_batch = np.random.randn(n + 1, 1)
    theta_sgd = np.random.randn(n + 1, 1)
    theta_mbgd = np.random.randn(n + 1, 1)

    errors = {"batch": [], "sgd": [], "minibatch": []}

    for epoch in range(n_epochs):
        # ---- Batch GD ----
        grad_batch = 2 / m * X_train_bias.T.dot(X_train_bias.dot(theta_batch) - y_train)
        theta_batch -= eta * grad_batch

        # ---- SGD ----
        i = np.random.randint(m)
        xi = X_train_bias[i:i + 1]
        yi = y_train[i:i + 1]
        eta_epoch = eta / (1 + decay * epoch) # Decaimiento de la tasa de aprendizaje
        grad_sgd = 2 * xi.T.dot(xi.dot(theta_sgd) - yi)
        theta_sgd -= eta_epoch * grad_sgd

        # ---- Mini-Batch GD ----
        indices = np.random.randint(m, size=batch_size)
        xi = X_train_bias[indices]
        yi = y_train[indices]
        grad_mbgd = 2 / batch_size * xi.T.dot(xi.dot(theta_mbgd) - yi)
        theta_mbgd -= eta * grad_mbgd

        # Calcular errores en test
        y_pred_batch = X_test_bias.dot(theta_batch)
        y_pred_sgd = X_test_bias.dot(theta_sgd)
        y_pred_mbgd = X_test_bias.dot(theta_mbgd)

        errors["batch"].append(mean_squared_error(y_test, y_pred_batch))
        errors["sgd"].append(mean_squared_error(y_test, y_pred_sgd))
        errors["minibatch"].append(mean_squared_error(y_test, y_pred_mbgd))

    # Graficar por cada eta
    plt.figure()
    plt.plot(range(1, n_epochs + 1), errors["batch"], label="Batch GD")
    plt.plot(range(1, n_epochs + 1), errors["sgd"], label="SGD")
    plt.plot(range(1, n_epochs + 1), errors["minibatch"], label="Mini-Batch GD")
    plt.xlabel("Épocas")
    plt.ylabel("Error (MSE en test)")
    plt.title(f"Comparación de Algoritmos (eta = {eta})")
    plt.legend()
    plt.show()
```

**Observaciones:**

eta = 0.001 (Tasa óptima): parece ser la tasa de aprendizaje ideal. Todos los modelos convergen de manera estable. El error de Batch GD y Mini-Batch GD desciende de manera muy parecida, mientras que el error de SGD desciende de forma estable hasta un valor bajo.

eta = 0.0001 (Tasa baja): el entrenamiento se vuelve muy lento. Aunque el error desciende, el cambio es mínimo en las 50 épocas probadas. El modelo necesitaría muchas más épocas para alcanzar el mismo error que con una tasa de 0.001.

eta = 1e-05 (Tasa muy baja): Con esta tasa de aprendizaje, el modelo prácticamente no aprende. Las curvas de error son casi planas, lo que indica que el algoritmo avanza de forma demasiado lenta para hacer una diferencia notable en el rendimiento.

**De Lasso y Ridge**

```python
# --- Código para Ridge con el nuevo rango ---
alphas_nuevos = [0.01, 0.05, 0.1, 0.5, 1, 5, 10, 50]

resultados_ridge_nuevos = []
for alpha in alphas_nuevos:
    ridge_model = Ridge(alpha=alpha)
    ridge_model.fit(X_train, y_train)

    # Evaluación y guardado de resultados...
    # (El resto del código es el mismo que ya tenías)
    y_pred_train = ridge_model.predict(X_train)
    y_pred_test = ridge_model.predict(X_test)

    resultados_ridge_nuevos.append({
        'Modelo': 'Ridge',
        'Alpha': alpha,
        'Conjunto': 'train',
        'R²': r2_score(y_train, y_pred_train),
        'MSE': mean_squared_error(y_train, y_pred_train)
    })
    resultados_ridge_nuevos.append({
        'Modelo': 'Ridge',
        'Alpha': alpha,
        'Conjunto': 'test',
        'R²': r2_score(y_test, y_pred_test),
        'MSE': mean_squared_error(y_test, y_pred_test)
    })

df_resultados_ridge_nuevos = pd.DataFrame(resultados_ridge_nuevos)
print(df_resultados_ridge_nuevos)
```

```python
# --- Código para Lasso con el nuevo rango ---
resultados_lasso_nuevos = []
for alpha in alphas_nuevos:
    lasso_model = Lasso(alpha=alpha)
    lasso_model.fit(X_train, y_train)
    
    # Evaluación y guardado de resultados...
    # (El resto del código es el mismo que ya tenías)
    y_pred_train = lasso_model.predict(X_train)
    y_pred_test = lasso_model.predict(X_test)

    resultados_lasso_nuevos.append({
        'Modelo': 'Lasso',
        'Alpha': alpha,
        'Conjunto': 'train',
        'R²': r2_score(y_train, y_pred_train),
        'MSE': mean_squared_error(y_train, y_pred_train)
    })
    resultados_lasso_nuevos.append({
        'Modelo': 'Lasso',
        'Alpha': alpha,
        'Conjunto': 'test',
        'R²': r2_score(y_test, y_pred_test),
        'MSE': mean_squared_error(y_test, y_pred_test)
    })

df_resultados_lasso_nuevos = pd.DataFrame(resultados_lasso_nuevos)
print(df_resultados_lasso_nuevos)
```

Es decir, el modelo no tiene un problema significativo de multicolinealidad que Ridge pudiera corregir. 

**Observaciones de Lasso:**

Para valores de alpha entre 0.01 y 0.10, el rendimiento es excelente. El R² se mantiene alto, cerca de 0.79, y el MSE es bajo. Esto demuestra que una penalización pequeña o nula no afecta el rendimiento.

Por otro lado, con valores de alpha de 5.00, el R² de prueba se desploma a 0.519545, y el MSE se dispara a 43.228545. Esto confirma que la penalización comienza a ser perjudicial.

Finalmente, con alpha de 10.00 y 50.00, el modelo falla completamente. El R² de prueba es casi 0 o incluso negativo, lo que significa que el modelo es inútil para hacer predicciones, es decir, hay un subajuste severo.
